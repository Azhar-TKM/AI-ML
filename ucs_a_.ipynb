```python

import heapq

def uniform_cost_search(graph, start, goal):
    pq = [(0, start, [])]
    visited = set()

    while pq:
        cost, node, path = heapq.heappop(pq)

        if node in visited:
            continue

        visited.add(node)
        path = path + [node]

        if node == goal:
            return path, cost

        for neighbor, weight in graph.get(node, []):
            if neighbor not in visited:
                heapq.heappush(pq, (cost + weight, neighbor, path))

    return None, None


graph = {}

edges = int(input("Enter number of edges: "))

print("Enter edges in the format:")
print("Source Destination Cost")

for _ in range(edges):
    u, v, cost = input().split()
    cost = int(cost)

    if u not in graph:
        graph[u] = []
    graph[u].append((v, cost))


start = input("Enter Start Node: ")
goal = input("Enter Goal Node: ")

path, cost = uniform_cost_search(graph, start, goal)

if path:
    print("\nShortest Path:", " -> ".join(path))
    print("Total Cost:", cost)
else:
    print("No path found.")
```

```python
import heapq

def a_star(graph, heuristic, start, goal):
    pq = [(heuristic[start], 0, start, [])]
    visited = set()

    while pq:
        f, g, node, path = heapq.heappop(pq)

        if node in visited:
            continue

        visited.add(node)
        path = path + [node]

        if node == goal:
            return path, g

        for neighbor, cost in graph.get(node, []):
            if neighbor not in visited:
                new_g = g + cost
                new_f = new_g + heuristic[neighbor]
                heapq.heappush(pq, (new_f, new_g, neighbor, path))

    return None, None


graph = {}

edges = int(input("Enter number of edges: "))

print("Enter edges (Source Destination Cost):")

for _ in range(edges):
    u, v, cost = input().split()
    cost = int(cost)

    if u not in graph:
        graph[u] = []

    graph[u].append((v, cost))

nodes = int(input("Enter number of heuristic values: "))

heuristic = {}

print("Enter Node and Heuristic Value:")

for _ in range(nodes):
    node, h = input().split()
    heuristic[node] = int(h)

start = input("Enter Start Node: ")
goal = input("Enter Goal Node: ")

path, cost = a_star(graph, heuristic, start, goal)

if path:
    print("\nShortest Path:", " -> ".join(path))
    print("Total Cost:", cost)
else:
    print("No path found.")
```
